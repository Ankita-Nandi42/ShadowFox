```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")

df = pd.read_csv("C:/Users/nandi/Downloads/delhiaqi.csv")

print(df.head())
print(df.info())
```

```python
print(df.isnull().sum())

df.drop_duplicates(inplace=True)

df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
df.describe()

df.corr(numeric_only=True)
```

```python
print("\n mean")
print(df.mean(numeric_only=True))

print("\n median")
print(df.median(numeric_only=True))

print("\n mode")
print(df.std(numeric_only=True))

print("\nStandard Deviation")
print(df.std(numeric_only=True))

print("\nCorrelation Matrix")
print(df.corr(numeric_only=True))

print("\nAnalysis Completed Successfully")
```

```python
pollutants = ["co","no","no2","o3","so2","pm2_5","pm10","nh3"]
plt.figure(figsize=(10,6))
sns.barplot(x=pollutants,
            y=df.mean(numeric_only=True))

plt.title("Average Pollutant Levels")
plt.xlabel("Pollutants")
plt.ylabel("Average Concentration")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["pm2_5"], bins=25, kde=True)
plt.title("Distribution of PM2.5")
plt.xlabel("PM2.5")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x="pm2_5",
                y="pm10",
                data=df)

plt.title("PM2.5 vs PM10")
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.boxplot(y=df["pm2_5"])
plt.title("Box Plot of PM2.5")
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(df[pollutants].corr(),
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# # Load dataset
# df = pd.read_csv("delhiaqi.csv")

# Convert date column
df["date"] = pd.to_datetime(df["date"])

# Plot PM2.5 over time
plt.figure(figsize=(12,6))
plt.plot(df["date"], df["pm2_5"], color="blue", linewidth=2)

plt.title("PM2.5 Levels Over Time")
plt.xlabel("Date")
plt.ylabel("PM2.5 (µg/m³)")
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# # Load dataset
# df = pd.read_csv("delhiaqi.csv")

# Calculate average pollutant concentrations
pollutants = ["pm2_5", "pm10", "no2", "so2", "o3"]

avg_values = df[pollutants].mean()

# WHO guideline values (µg/m³)
who_limits = {
    "pm2_5": 15,
    "pm10": 45,
    "no2": 25,
    "so2": 40,
    "o3": 100
}

# Plot
plt.figure(figsize=(10,6))

plt.bar(avg_values.index, avg_values.values,
        color=["red","orange","green","blue","purple"])

plt.plot(avg_values.index,
         list(who_limits.values()),
         color="black",
         marker="o",
         linewidth=2,
         label="WHO Safe Limit")

plt.title("Average Pollutant Levels vs WHO Guidelines")
plt.xlabel("Pollutants")
plt.ylabel("Concentration (µg/m³)")
plt.legend()
plt.grid(axis="y")

plt.show()

print("\n========== Air Quality Improvement Strategies ==========\n")

if df["pm2_5"].mean() > 15:
    print("• Reduce PM2.5 by controlling vehicle emissions and industrial pollution.")

if df["pm10"].mean() > 45:
    print("• Control construction dust and improve road cleaning.")

if df["no2"].mean() > 25:
    print("• To control no2: Promote electric vehicles and cleaner fuels.")

if df["so2"].mean() > 40:
    print("• Reduce coal burning and industrial emissions.")

if df["o3"].mean() <= 100:
    print("• Air Quality Status: O₃ is within the WHO safe limit.")
else:
    print("• Air Quality Status: O₃ exceeds the WHO safe limit.")

print("• Increase urban green cover.")
print("• Strengthen air quality monitoring stations.")
print("• Encourage the use of public transportation.")
print("• Conduct public awareness campaigns on air pollution.")
print("• Issue health advisories during high pollution periods.")
```
